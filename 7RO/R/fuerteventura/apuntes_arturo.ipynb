```python
%%R
moderado <- subset(df, temp >= 20 & temp <=25 & ws >= 5 & ws <=10)
head(moderado,50)
```

```python
%%R
vis_10 <- subset(df, visibility >= 10000 & !is.na(temp))
head(vis_10,50)
```

```python
%%R
#Lo primero es sacar la media
temp_media = mean(df$temp, na.rm = TRUE)

#Subconjunto con temperatura menor que la media y de ahi hay que seleccionar fecha, temp y ws
dias_temp_menor_media = subset(df, temp < temp_media, select= c(temp, Date, ws))

#Ahora ordenamos los datos para ello primero se genera el vector de las posiciones ordenadas
posiciones_mas_frias <- order(dias_temp_menor_media$temp, na.last = TRUE)

#Ahora creamos el nuevos data frame usando las posiciones para crear las filas
dias_temp_menor_media_ordenados <- dias_temp_menor_media[posiciones_mas_frias,]
head(dias_temp_menor_media_ordenados, 50)

```
